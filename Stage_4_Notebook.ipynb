```python
from pathlib import Path

# Adjust this if your Drive folder name is different
COLAB_NOTEBOOKS = Path("/content/drive/MyDrive/Colab Notebooks")
TEMPLATE_ROOT = COLAB_NOTEBOOKS / "ECS170_Spring_2026_Source_Code_Template"

required_paths = {
    "Template folder": TEMPLATE_ROOT,
    "Stage 4 classification data": TEMPLATE_ROOT / "data/stage_4_data/text_classification/train/pos",
    "Stage 4 generation data": TEMPLATE_ROOT / "data/stage_4_data/text_generation/data",
    "Stage 4 code folder": TEMPLATE_ROOT / "local_code/stage_4_code",
    "Stage 4 scripts folder": TEMPLATE_ROOT / "script/stage_4_script",
    "Stage 4 results folder": TEMPLATE_ROOT / "result/stage_4_result",
}

required_files = [
    TEMPLATE_ROOT / "local_code/stage_4_code/Dataset_Loader.py",
    TEMPLATE_ROOT / "local_code/stage_4_code/Method_RNN.py",
    TEMPLATE_ROOT / "script/stage_4_script/script_gru_classification.py",
    TEMPLATE_ROOT / "script/stage_4_script/script_lstm_generation.py",
]

print("Checking zip extraction...\n")
all_ok = True

for label, path in required_paths.items():
    ok = path.exists()
    status = "OK" if ok else "MISSING"
    print(f"[{status}] {label}: {path}")
    all_ok = all_ok and ok

print("\nChecking key Stage 4 files...\n")
for path in required_files:
    ok = path.is_file()
    status = "OK" if ok else "MISSING"
    print(f"[{status}] {path.name}")
    all_ok = all_ok and ok

# Optional counts to confirm data is not empty
if (TEMPLATE_ROOT / "data/stage_4_data/text_classification/train/pos").exists():
    pos_count = len(list((TEMPLATE_ROOT / "data/stage_4_data/text_classification/train/pos").glob("*.txt")))
    neg_count = len(list((TEMPLATE_ROOT / "data/stage_4_data/text_classification/train/neg").glob("*.txt")))
    print(f"\nClassification train files: pos={pos_count}, neg={neg_count}")

if (TEMPLATE_ROOT / "data/stage_4_data/text_generation/data").exists():
    joke_lines = sum(1 for _ in open(TEMPLATE_ROOT / "data/stage_4_data/text_generation/data", encoding="utf-8")) - 1
    print(f"Generation jokes (rows): {joke_lines}")

print("\n" + ("Zip extraction looks GOOD. You can start training." if all_ok else "Something is missing. Re-upload or re-unzip the zip file."))
```
